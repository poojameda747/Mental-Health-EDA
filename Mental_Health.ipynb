# **Project Summary -**

This project performs an Exploratory Data Analysis (EDA) on the Mental Health in Tech Survey dataset collected in 2014 by Open Sourcing Mental Illness (OSMI). The dataset captures attitudes towards mental health and frequency of mental health disorders among 1,259 tech industry employees across 48 countries.

The primary goal is to uncover patterns related to mental health treatment, workplace support structures, stigma, and employee behaviour. We explore how demographics (age, gender, country), company characteristics (size, tech focus, remote work), and employer policies (benefits, wellness programs, leave policies) influence mental health outcomes in the workplace.

The analysis follows the UBM framework — Univariate, Bivariate, and Multivariate Analysis — to systematically investigate variable distributions, relationships, and combined effects. Data cleaning addresses issues like invalid age values and inconsistent gender labels. Over 15 meaningful charts are produced to surface actionable insights.

Key findings include: roughly half of respondents have sought mental health treatment; family history is a strong predictor of treatment-seeking; many employees are uncertain about employer benefits and anonymity protections; and stigma around discussing mental health — especially in interviews — remains high. These insights can guide HR and leadership teams to design better mental health support frameworks, reduce workplace stigma, and improve employee wellness outcomes.

## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.

# --- 1. Drop 'comments' column (87% missing, not useful for EDA) ---
df.drop(columns=['comments', 'Timestamp', 'state'], inplace=True)

# --- 2. Fix Age: remove invalid ages (keep 18–65) ---
df = df[(df['Age'] >= 18) & (df['Age'] <= 65)]
print(f"Rows after age filter: {len(df)}")

# --- 3. Standardize Gender ---
def clean_gender(g):
    g = str(g).strip().lower()
    if g in ['male', 'm', 'man', 'cis male', 'male (cis)', 'maile', 'mal', 'male-ish', 'make', 'msle', 'mail', 'malr', 'cis man']:
        return 'Male'
    elif g in ['female', 'f', 'woman', 'cis female', 'femake', 'female (cis)', 'femail', 'cis-female/femme', 'female ']:
        return 'Female'
    else:
        return 'Other/Non-binary'

df['Gender'] = df['Gender'].apply(clean_gender)
print("Gender counts after cleaning:")
print(df['Gender'].value_counts())

# --- 4. Fill missing values ---
df['self_employed'] = df['self_employed'].fillna('No')
df['work_interfere'] = df['work_interfere'].fillna("Don't know")

# --- 5. Encode binary columns numerically for correlation ---
binary_map = {'Yes': 1, 'No': 0}
for col in ['family_history', 'treatment', 'remote_work', 'tech_company', 'self_employed', 'obs_consequence']:
    df[col] = df[col].map(binary_map)

# --- 6. Encode work_interfere ordinal ---
work_map = {'Never': 0, 'Rarely': 1, 'Sometimes': 2, 'Often': 3, "Don't know": -1}
df['work_interfere_enc'] = df['work_interfere'].map(work_map)

print("\nData wrangling complete!")
print(f"Final dataset shape: {df.shape}")
df.head()
```

### What all manipulations have you done and insights you found?

The following transformations were applied to make the dataset analysis-ready:

1. **Dropped irrelevant columns**: `comments` (87% missing), `Timestamp` (not useful), and `state` (only meaningful for US analysis and too sparse).
2. **Age cleaning**: Removed rows with unrealistic ages (< 18 or > 65). This eliminated extreme outliers like negative values and values in the billions.
3. **Gender standardization**: Consolidated 49 inconsistent gender labels into three clean categories: `Male`, `Female`, and `Other/Non-binary`.
4. **Missing value imputation**: Filled `self_employed` with 'No' (most common) and `work_interfere` with "Don't know" (acknowledging true uncertainty).
5. **Binary encoding**: Mapped Yes/No columns to 1/0 for correlation analysis.
6. **Ordinal encoding**: Created a numeric `work_interfere_enc` column mapping interference levels from Never (0) to Often (3).

These steps ensured the data is clean, consistent, and ready for meaningful visualization and analysis.
